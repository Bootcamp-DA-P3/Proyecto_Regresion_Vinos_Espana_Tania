## 1. Resumen Ejecutivo

El presente informe detalla la evaluación cuantitativa y cualitativa del modelo de *Machine Learning* desarrollado para predecir el precio ($€$) de los vinos en el mercado español. A partir del análisis exploratorio y del ajuste de hiperparámetros, se ha obtenido un modelo basado en **Random Forest Regressor** estructurado para equilibrar capacidad predictiva y generalización.

---

## 2. Métricas de Rendimiento de Regresión

Para evaluar la capacidad predictiva del modelo y verificar la presencia o ausencia de *overfitting* (sobreajuste), se comparan las métricas clave en los conjuntos de **Entrenamiento (Train)** y **Prueba (Test)**.

### 2.1 Tabulación de Métricas

| Métrica | Conjunto Train | Conjunto Test | Brecha / Variación |
| :--- | :---: | :---: | :---: |
| **$R^2$ (Coeficiente de Determinación)** | 0.5012 | 0.4797 | $4.29\%$ |
| **RMSE (Root Mean Squared Error)** | $114.85\text{ €}$ | $117.46\text{ €}$ | $+2.27\%$ |
| **MAE (Mean Absolute Error)** | $31.20\text{ €}$ | $32.85\text{ €}$ | $+5.28\%$ |

### 2.2 Diagnóstico de Sobreajuste (Overfitting)

* **Análisis del $R^2$:** La diferencia relativa de rendimiento entre el conjunto de entrenamiento ($R^2 = 0.5012$) y el conjunto de prueba ($R^2 = 0.4797$) es del **$4.29\%$**.
* **Conclusión:** Al ser la brecha inferior al **$5\%$**, el modelo demuestra una sólida capacidad de generalización con datos no vistos, descartando problemas severos de memoria o sobreajuste (*overfitting*).

## 3. Importancia de las Características (*Feature Importance*)

El análisis de la relevancia de variables generada por el ensemble de árboles arroja la siguiente distribución de pesos descriptivos:

```
Puntuación (Rating)     [########################################] 42.3%
Num. Reseñas (Reviews)   [============================] 28.1%
Cuerpo (Body)            [==============] 14.5%
Acidez (Acidity)         [=======] 7.2%
Año de Cosecha (Year)    [====] 4.6%
Otras Categorías         [==] 3.3%
```

### Principales Hallazgos:
1. **Puntuación (`rating`):** Es la variable de mayor peso intrínseco ($42.3\%$). La percepción de calidad de los evaluadores/usuarios es el determinante directo principal del precio final.
2. **Volumen de Opiniones (`num_reviews`):** Representa un $28.1\%$ de la importancia, reflejando que la popularidad y el reconocimiento del mercado influyen positivamente en la valoración económica.
3. **Atributos Organolépticos (`body` y `acidity`):** El cuerpo ($14.5\%$) y la acidez ($7.2\%$) aportan un nivel secundario de diferenciación técnica dentro de las características estructurales del vino.

## 4. Análisis de Residuos e Interpretación Gráfica

### 4.1 Residuos ($\text{Real} - \text{Predicción}$)
El análisis de los errores residuales permite identificar comportamientos atípicos en las predicciones:

* **Homocedasticidad en Rango Medio/Bajo:** Para vinos con precios situados entre $0\text{ €}$ y $150\text{ €}$, los residuos se distribuyen de manera simétrica alrededor del valor $0$, lo que confirma que el modelo realiza estimaciones estables y precisas dentro del segmento comercial estándar.
* **Sesgo en Vinos de Alta Gama (*Outliers*):** En botellas de colección o edición limitada que superan los $500\text{ €}$, el modelo tiende a **subestimar el precio**. Esto es una consecuencia esperada debido a la baja frecuencia de datos en valores extremos y al efecto de compresión de varianza típico de los métodos basados en árboles de decisión.

### 4.2 Comportamiento Predicción vs. Valores Reales
Al representar la gráfica de **Valores Reales vs. Valores Predichos** frente a la línea ideal de $45^\circ$ ($y = x$):
* Se observa una alineación consistente a lo largo de la diagonal en la mayor densidad de datos.
* La dispersión aumenta progresivamente a medida que el precio real se incrementa, recomendando aplicar transformaciones logarítmicas sobre la variable objetivo en futuras iteraciones si se busca reducir el error en el segmento *premium*.

---

## 5. Recomendaciones Técnicas e Itinerario Futuro

1. **Transformación Logarítmica:** Aplicar $\log(y)$ sobre el precio para estabilizar la varianza de los valores extremos (*outliers*).
2. **Ingeniería de Características Adicionales:** Explorar la interacción entre la bodega (`winery`) y la denominación de origen (`region`) para capturar la reputación de marca.
3. **Algoritmos Basados en Boosting:** Probar algoritmos como *XGBoost* o *LightGBM* con regularización $L1/L2$ para evaluar posibles ganancias marginales en el métrico $R^2$.
